# **File Access in Google Colab**

## Why Direct Access to Local Files is Not Possible

Google Colab is a hosted service that runs on Google's servers. This means the Python runtime environment you are using is not running directly on your local machine. Consequently, the code executed in a Colab notebook does not have direct access to your local file system for security and technical reasons.

## Methods for Accessing Data in Colab

To work with data in Google Colab, you need to bring your data into the Colab environment. This notebook will demonstrate several common methods for accessing data:

1.  **Uploading Files from Your Local Machine:** This method allows you to upload individual files or folders directly to the Colab environment for temporary use.
2.  **Mounting Google Drive:** You can connect your Google Drive to Colab, providing persistent access to files stored in your Drive.
3.  **Downloading from URLs:** You can download files directly from the web using their URLs.


Each of these methods has its advantages and is suitable for different scenarios. The following sections will detail how to use each method.



## 1. **Uploading Files from Your Local Machine**



```python
from google.colab import files

print("Please select files to upload from your local machine.")
uploaded = files.upload()
print("File upload complete.")

# now you can use the file called "upoaded"...
# each time you open and close this notebook, you have to repeat this step
```

## 2.  **Mounting Google Drive**

This is the best method. It gives you access to your google drive and it behaves like normal folders.



```python
from google.colab import drive

print("Mounting Google Drive...")
drive.mount('/content/drive')
print("Google Drive mounted successfully!")
```

## 3.  **Downloading from URLs**


```python
import requests

url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
response = requests.get(url)

if response.status_code == 200:
    file_name = url.split('/')[-1]
    with open(file_name, 'wb') as f:
        f.write(response.content)
    print(f"File downloaded successfully: {file_name}")
else:
    print(f"Failed to download file from {url}. Status code: {response.status_code}")
```

## Summary of File Access Methods in Google Colab

We've explored several methods for accessing data in your Google Colab environment:

*   **Uploading Files from Your Local Machine:** This is useful for small files or when you need to quickly bring data into Colab for temporary use. Remember that these files are not persistent and will be lost when the Colab session ends.

*   **Mounting Google Drive:** This method provides persistent storage and is ideal for larger datasets or when you need to access files that you'll be using across multiple Colab sessions. It integrates seamlessly with your Google Drive.

*   **Downloading from URLs:** This is a convenient way to access publicly available data directly from the web without needing to download it to your local machine first.

Choosing the right method depends on the size and location of your data, as well as whether you need persistent storage. By utilizing these techniques, you can effectively work with your data within the Google Colab environment.
