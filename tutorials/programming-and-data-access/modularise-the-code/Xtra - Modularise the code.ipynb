# Python Functions and Modules

As a program grows, it becomes useful to move related functions into a separate Python file. Such a file is called a **module**. You can import it into notebooks and other programs instead of copying the same code repeatedly.

A related but distinct task is keeping **credentials**—passwords, API keys and tokens—out of code that you share. A separate file helps organise information, but it only improves security if you protect that file from publication. This short exercise covers both ideas.

## 1. Import reusable functions

The supplied `functions.py` is a module containing two functions, `sayHello()` and `addTwo(x)`. A function takes inputs (arguments), performs an operation and may **return** a result. Keep `functions.py` in the same folder as this notebook, then run the cells below.

```python
import functions

print(functions.sayHello())
print(functions.addTwo(3))
```

Open `functions.py` and read the two definitions. Change the text returned by `sayHello()` in your own copy, save the module, then **restart the notebook kernel** and run the import again. Restarting is a simple way to load your changes; Python usually caches already imported modules. Notice how the notebook can use functions without containing their implementation.

## 2. Separate example credentials from your main program

The folder includes `Passwords.py` and `passwords.csv` containing **dummy values only**. They illustrate two ways of storing configuration outside the main notebook. They are committed here only because they are fictional examples—**never replace them with genuine credentials in this repository**.

You can import a Python module in the same folder just as you imported `functions.py`. The next cell accesses an example value without displaying it.

```python
import Passwords as example_credentials

dummy_key = example_credentials.password_flickr
print('Loaded a dummy credential:', isinstance(dummy_key, str))
```

A CSV file can also hold configuration, but storing passwords in CSV or Python files does **not** encrypt or otherwise protect them. For this introductory exercise, you do not need to use the supplied CSV. Do not print, display or save real secrets in notebook output: notebooks can store their execution results and those results may be committed or shared.

## 3. Keep private configuration out of Git

If a project needs a local file with real credentials, give it a private name such as `private_passwords.py` or `.env` and add that name to `.gitignore` **before creating or committing it**. This folder includes a sample [`.gitignore`](.gitignore) to show the principle. Do not put actual passwords in a file tracked by Git.

**Important:** `.gitignore` applies to files Git is not already tracking. It does not erase a file from existing commits or remove it from version history. The supplied `Passwords.py` and `passwords.csv` remain tracked because they contain only placeholders.

A separate private file is useful for local experiments, but an **environment variable** is often a better way to supply a secret to a program without embedding it in source code.

## 4. Read a credential from an environment variable

An **environment variable** is a named value available to a running process. You can configure one outside the notebook and read it in Python using the standard `os` module. The following code deliberately does not reveal the value. If it has not been configured, you will see a harmless message.

```python
import os

api_key = os.getenv('RC15_DEMO_API_KEY')
if api_key:
    print('The environment variable is available.')
else:
    print('RC15_DEMO_API_KEY has not been set.')
```

To test this, set an environment variable with a **fictional value**, then restart your notebook environment so it inherits that variable. On Windows PowerShell, for example, run `$env:RC15_DEMO_API_KEY = 'dummy-value'` before starting Jupyter from the same terminal. On macOS/Linux, use `export RC15_DEMO_API_KEY='dummy-value'` before launching Jupyter from that shell. These are session-level examples, not a secure secret-management service. Avoid entering genuine secrets into a terminal command if your shell history, process environment or computer is not adequately protected.

If an application requires a real credential, consult its documentation for safe storage, permissions and rotation. A password manager, operating-system credential store or appropriately configured secret manager may be more suitable than plain-text files. Keep real keys out of shared code, notebooks, screenshots, logs and Git history.
