# File Access in Google Colab

Google Colab runs Python on a remote computer rather than directly on your laptop. To analyse a CSV, image or other dataset, you first need to make the file available to that remote environment.

This notebook demonstrates three ways to do that: **upload a local file**, **mount Google Drive**, or **download a file from a URL**. Run each code cell and look for the resulting files in Colab's **Files** panel. Remember that files stored only in the runtime may disappear when the runtime is reset or disconnected.


## 1. Upload a file from your computer

Use this for a quick experiment with a file you already have. Colab opens a file picker and copies the selected file into the current runtime. The variable `uploaded` contains the uploaded filenames and their contents; it is not itself a filename.


```python
from google.colab import files

print("Please select files to upload from your local machine.")
uploaded = files.upload()
print("File upload complete.")

# now you can use the file called "uploaded"...
# each time you open and close this notebook, you usually need to upload the file again in a new runtime session
```

## 2. Mount Google Drive

Use this when files need to remain available across sessions. Authorise the connection and browse your files under `/content/drive/`. Mounting gives the notebook access to your Drive, so review notebook code before granting permission.


```python
from google.colab import drive

print("Mounting Google Drive...")
drive.mount('/content/drive')
print("Google Drive mounted successfully!")
```

## 3. Download a file from a URL

Use this for a publicly accessible online dataset. The example downloads a CSV into the runtime's current working directory. A successful download does not automatically save it to your Google Drive.


```python
import requests

url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
response = requests.get(url)

if response.status_code == 200:
    file_name = url.split('/')[-1]
    with open(file_name, 'wb') as f:
        f.write(response.content)
    print(f"File downloaded successfully: {file_name}")
else:
    print(f"Failed to download file from {url}. Status code: {response.status_code}")
```
